```python
## 3 Display all employees sorted by salary from highest to lowest. Show only Name, Department, and Salary. 
```

```python
df = df.sort_values(by ="Salary",ascending=False)
df[["Name","Department","Salary"]]
```

```python
## 4 Find the mean, maximum, minimum, and median salary. Also find the average age of all employees.
```

```python
print(df["Salary"].mean())
print(df["Salary"].max())
print(df["Salary"].min())
print(df["Salary"].median())
print(df["Age"].mean())
```
