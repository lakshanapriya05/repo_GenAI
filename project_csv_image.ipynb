```python
import os

```

```python
import pandas as pd

```

```python
import matplotlib.pyplot as plt
```

```python
path = "C:\\Users\\Jeyaprakash\\Desktop\\"

df = pd.read_csv(os.path.join(path, "proj_csv.csv"))

fig, ax = plt.subplots(figsize=(10, 6))

colors = plt.cm.Greens(df["Frequency"] / df["Frequency"].max())

ax.barh(df["Ip"],df["Frequency"],color=colors,edgecolor="green")

ax.set_title("IP Address Frequency",color="darkgreen")

ax.set_xlabel("Frequency (Number of Requests)")
ax.set_ylabel("IP Address")

# Display frequency values
for i, value in enumerate(df["Frequency"]):
    ax.text(value, i, str(value), va="center", ha="left")

plt.tight_layout()

# Save chart as image
plt.savefig(os.path.join(path, "ip_frequency.png"), dpi=300, bbox_inches="tight")

plt.show()
```
