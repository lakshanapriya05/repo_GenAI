**STRING FORMATING** 

* %d = integer
* %f = float with default 6 decimal value
* %s = string
* %.2f = float with 2 decimal value
* %.10f = float with 10 decimal place

```python
name = "Priya"
age = 24
height = 144.44

print("%s is %d year old and height is %.2f" %(name,age,height))
```

```python
length = 123.5050
weight = 150.5000
height = 7050.1500

print("length = %f\nwweight = %.2f\nheight = %.8f" %(length,weight,height))
```

```python
quantity = 30
itemno = 50
price =  250

print(f" I want {quantity} piece of item no{itemno} for ${price} dollars.")
```
