**DICTIONARY FUNCTIONS**


```python
mydict ={
    "Name" : "Lakshana",
    "City" : "Chennai",
   "Course": "Python",
       500 : 50.55,
}

print(mydict,type(mydict))
print(mydict["Name"])
print(mydict["Course"])

mydict["Name"]="Pooja"
mydict["age"]= 24
print(mydict)
```

```python
mydict ={
    "Name" : "Lakshana",
    "City" : "Chennai",
   "Course": "Python",
       500 : 50.55,
}
print(mydict)
print(list(mydict.keys()))
print()

for each in mydict.keys():
    print(each)
 
```

```python
# values()

mydict ={
    "Name" : "Lakshana",
    "City" : "Chennai",
   "Course": "Python",
       500 : 50.55,
}
print(mydict)
print(list(mydict.values()))
print()
for each in mydict.values():
    print(each)
```

```python
# items both key and the value should come in output

mydict ={
    "Name" : "Lakshana",
    "City" : "Chennai",
   "Course": "Python",
       500 : 50.55,
}

print(mydict)
print(mydict.items())
print()

for each in mydict.items():
    print(f"{each[0]}\t{each[1]}")

```

```python
# list = pop() - dlt the last value
# list = pop(index)-particularly delete the mentioned index

mylist = [100,30.55,True,30.55,"Priya",100,"Hi"]
print(mylist)
mylist.pop()
print(mylist)
mylist.pop(1)
print(mylist)
```

```python
# set = pop() dlt the duplicates

myset =  {100,30.55,True,30.55,"Priya",100,"Hi"}
print(myset)
myset.pop()
print(myset)
```

```python
# pop(key)

mydict ={
    "Name" : "Lakshana",
    "City" : "Chennai",
   "Course": "Python",
       500 : 50.55,
}

print(mydict)
mydict.pop("City")
print(mydict)
```

```python
#pop(item) - dlt the last function

mydict ={
    "Name" : "Lakshana",
    "City" : "Chennai",
   "Course": "Python",
       500 : 50.55,
}

print(mydict)
mydict.popitem()
print(mydict)
```

```python
# clear()- will return empty string

mydict ={
    "Name" : "Lakshana",
    "City" : "Chennai",
   "Course": "Python",
       500 : 50.55,
}

print(mydict)
mydict.clear()
print(mydict)
```

```python
#update

mydict ={
    "Name" : "Lakshana",
    "City" : "Chennai",
   "Course": "Python",
       500 : 50.55,
}

print(mydict)
newdit = {"Name" : "Pooja", "City": "Pune"}
mydict.update(newdit)
print(mydict)
```

```python
#get()

mydict ={
    "Name" : "Lakshana",
    "City" : "Chennai",
   "Course": "Python",
       500 : 50.55,
}

print(mydict)
print(mydict[500])
print(mydict.get(500))
print(mydict.get(600))
print("Hello world")

```

```python
# copy()

x = {"A":100,"B":200}
y = {"A":100,"B":200}

x["C"]=300
y["D"]=400

print(x,type(x),id(x))
print(y,type(y),id(y))
```

```python
# copy()- save in different location

x = {"A":100,"B":200}
y = x.copy()

x["C"]=300
y["D"]=400

print(x,type(x),id(x))
print(y,type(y),id(y))
```

**NESTED DICTIONARY**

```python
mydict = {
    "Name":"Lakshana",
    "city":["chennai","covai","Pune"],
    "Year":(2002,2026,2030)
}
print(mydict,type(mydict))
print(mydict.keys())
print(mydict.values())
print(mydict.get("Name"))
print(mydict.get("city"))
print(mydict.get("city")[0])
print(mydict.get("Year")[2])
print('------------------------')
print(mydict.get("city")[0][3:])
```

```python

 x = {
    "menu": {
        "id": "file",
        "value":"File",
        "popup": {
            "menuitem":[
                {"value": "New","onclick": "CreateNewDoc()"},
                            {"value": "Open","onclick": "OpenDoc()"},
                            {"value": "Close","onclick":"CloseDoc()"} 
                        
            ]
            
        }

    }
    
}

print(x,type(x))
print(x.keys())
print('--------------')

print(x.get("menu")['popup'])

print(x.get("menu")['popup']['menuitem'])
print(x.get("menu")['popup']['menuitem'][2])
print(x.get("menu")['popup']['menuitem'][1]["onclick"])
print(x.get("menu")['popup']['menuitem'][1]["onclick"])
    


 

 
```
