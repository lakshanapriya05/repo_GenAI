# **Scatter plot**

```python
import numpy as np
```

```python
x = np.array([1,2,3,4,5])
y = np.array([5,4,2,3,1])
plt.scatter(x,y)
plt.show();
```

```python
## customise the plot
```

```python
x = np.array([1,2,3,4,5])
y = np.array([5,4,2,3,1])
color = ["r","g","b","orange","violet"]

plt.scatter(x,y,c=color,marker="D",s=100)
plt.show();
```

# **Line Plot**

```python
x = np.array([1,2,3,4,5])
y = np.array([5,4,2,3,1])
plt.plot(x,y,c="b",marker="*",linestyle="dashed",ms=20)
plt.show();
```

```python
x = np.array([1,2,3,4,5])
y = np.array([5,4,2,3,1])
c= ["red","black","orange","violet","green"]

plt.plot(x,y,c="g",linestyle="dotted",marker="*",)
plt.scatter(x,y,color = c,marker="*",s=500)
plt.show();
```

# **Barchart**

# ** Pie chart**

```python
products = ["Laptop", "Mobile", "Tablet", "Watch", "TV"]
revenue = [120000, 200000, 80000, 60000, 150000]
exp=[0,0.1,0,0,0]
c = ["violet","indigo","cyan","pink","orange"]
plt.pie(revenue,labels=products,colors=c,explode=exp)
plt.show();
```

```python
experience = [1, 2, 3, 4, 5]
salary = [25000, 28000, 32000, 38000, 42000]
plt.title("Employee Experience vs Annual Salary")
plt.xlabel(experience)
plt.ylabel(salary)
plt.plot(experience,salary,c="b",marker="^",ms=10)
plt.show();
          
```

# **stock market**

```python
stockA = [1,2,3,4,5]
stockB = [3,4,5,2,1]
stockC = [5,2,3,4,1]
plt.plot(stockA)
plt.plot(stockB)
plt.plot(stockC)
plt.show();
```

```python
stockA = [1,2,3,4,5]
stockB = [3,4,5,2,1]
stockC = [5,2,3,4,1]
plt.plot(stockA)
plt.plot(stockB)
plt.plot(stockC)
plt.title("Stock Market")
plt.plot(stockA,label="stock1",marker="*")
plt.plot(stockB,label="stock2",marker="D")
plt.plot(stockC,label="stock3",marker="^")
plt.legend()
plt.show();
```

# *subplots comparing (age vs chol),(age vs thalach)*

```python
fig,ax=plt.subplots(nrows=2,ncols=1,figsize=(10,5))
```

```python
#subplots
fig,ax=plt.subplots(nrows=2,ncols=1,figsize=(10,5))
scatter = ax[0].scatter(df["age"],df["chol"],c=df["target"],cmap="winter")

#customise the plot

ax[0].set(title ="Age vs Cholestrol", xlabel="age",ylabel="cholestrol")
ax[0].legend(*scatter.legend_elements(),title="target")

# add mean line

ax[0].axhline(df["chol"].mean(),color="r",linestyle ="dotted")


scatter = ax[1].scatter(df["age"],df["thalach"],c=df["target"],cmap="coolwarm")
ax[1].set(title ="Age vs thalach", xlabel="age",ylabel="thalach")
ax[1].legend(*scatter.legend_elements(),title="target")
ax[1].axhline(df["thalach"].mean(),color="r",linestyle ="dotted")
plt.subplots_adjust(hspace= 0.4)
plt.show();


```
